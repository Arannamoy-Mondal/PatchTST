```python
from datetime import datetime, timezone, timedelta


symbol = "BTCUSD"       
timeframe = mt5.TIMEFRAME_H1

start = datetime(2013, 1, 1, tzinfo=timezone.utc)
end = datetime.now(timezone.utc)

frames = []
current = start

while current < end:
    next_date = min(
        current + timedelta(days=30),
        end
    )

    rates = mt5.copy_rates_range(
        symbol,
        timeframe,
        current,
        next_date
    )

    if rates is not None and len(rates) > 0:
        frames.append(pd.DataFrame(rates))
        print(current.date(), "to", next_date.date(), len(rates))
    else:
        print("No data:", current.date(), mt5.last_error())

    current = next_date + timedelta(seconds=1)

if not frames:
    raise RuntimeError("No data found")

df = pd.concat(frames, ignore_index=True)

df["time"] = pd.to_datetime(
    df["time"],
    unit="s",
    utc=True
)

df = (
    df.drop_duplicates("time")
      .sort_values("time")
      .reset_index(drop=True)
)

print("Total rows:", len(df))
print("First available:", df["time"].min())
print("Last available:", df["time"].max())

df.to_csv(
    f"{symbol}_H1_from_2013.csv",
    index=False
)

print(df.head(1),df.tail(1))
```
